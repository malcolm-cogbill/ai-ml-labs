#Part 1: CONCEPTUAL QUESTIONS

1. What differs data science from data analytics?

Data analytics focuses on understanding what happened and why. Data science focuses on predicting what will happen and creating data-driven solutions.

2. Between mean and median, what do you think is more sensitive to outliers? Why?

The mean is more sensitive to outliers because it incorporates the actual value of every observation. The median is more resistant to outliers because it only depends on the order of the data and the middle position.

3. Is correlation also causation? Why or why not?

No, correlation is not the same as causation. A correlation means that two variables tend to change together. A causal relationship means that one variable directly causes a change in the other.

For the age, let's say that our data is gathered on the last day of 2015.

**Labeling Age categories**

'Gen Z (0-24)': Ages from 0 to 24 (inclusive)

'Gen Y (25-39)': Ages from 25 to 39 (inclusive)

'Gen X (40-54)': Ages from 40 to 54 (inclusive)

'Older (55-59)': Ages from 55 to 59 (inclusive)

'Senior (60+)': Ages 60 and above

```python
import pandas as pd

# SuperStore data from local CSV file
data = pd.read_csv('/content/SuperStore-Data (2).csv')

# Data is gathered on last day of 2015
data['Age'] = 2015 - data['Year_Birth']

# Define the age categories and labels
bins = [0, 24, 39, 54, 59, data['Age'].max() + 1]
labels = ['Gen Z (0-24)', 'Gen Y (25-39)', 'Gen X (40-54)', 'Older (55-59)', 'Senior (60+)']

data['AgeCat'] = pd.cut(data['Age'], bins=bins, labels=labels, right=True, include_lowest=True)

display(data[['Year_Birth', 'Age', 'AgeCat']].head())
display(data['AgeCat'].value_counts().sort_index())

# Calculate Total Spending by summing relevant columns
data['TotalSpending'] = data['MntWines'] + data['MntFruits'] + data['MntMeatProducts'] + \
                        data['MntFishProducts'] + data['MntSweetProducts'] + data['MntGoldProds']

# Insight 1: Average Income by Age Category
print("\nInsight 1: Average Income by Age Category:")
display(data.groupby('AgeCat', observed=False)['Income'].mean().sort_index().round(2))

# Insight 2: Average Total Spending by Age Category
print("\nInsight 2: Average Total Spending by Age Category:")
display(data.groupby('AgeCat', observed=False)['TotalSpending'].mean().sort_index().round(2))

# Insight 3: Education Level Distribution by Age Category
print("\nInsight 3: Education Level Distribution by Age Category:")
education_distribution = data.groupby('AgeCat', observed=False)['Education'].value_counts(normalize=True).mul(100).rename('percentage').reset_index()
display(education_distribution)
```

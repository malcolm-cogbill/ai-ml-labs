```python
import pandas as pd

# Importing data from Kaggle
Data = 'Video_Games_Sales_as_at_22_Dec_2016.csv'

# Read CSV file in DataFrame
df = pd.read_csv(Data)

df.info()
```

```python
display(df.head())
```

```python
missing_values = df.isnull().sum()
display(missing_values)
```

```python
# Clean data
df['Publisher'] = df['Publisher'].fillna('Unknown')

# Change missing 'Year_of_Release' values to the median
df['Year_of_Release'] = df['Year_of_Release'].fillna(df['Year_of_Release'].median())
df['Year_of_Release'] = df['Year_of_Release'].astype(int)

# Drop columns with high number of missing values
columns_to_drop = ['Critic_Score', 'Critic_Count', 'User_Score', 'User_Count', 'Developer', 'Rating']
df.drop(columns_to_drop, axis=1, inplace=True, errors='ignore')

missing_values = df.isnull().sum()
display(missing_values)

df.info()
```

```python
# Drop rows where 'Name' or 'Genre' are missing
df.dropna(subset=['Name', 'Genre'], inplace=True)

missing_values_after_drop = df.isnull().sum()
display(missing_values_after_drop)

df.info()
```

```python
# Insight in top 10 games sold in 2016
top_sales = df.sort_values(by = 'Global_Sales', ascending = False).head(10)
top_sales
```

```python
# Get mean, median, and mode for all sales columns
sales_columns = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']

display(df[sales_columns].mean())

display(df[sales_columns].median())

display(df[sales_columns].mode())
```

```python
# Platform, Genre, and Publisher with most games released
list_columns = ['Platform', 'Genre', 'Publisher']

for col in list_columns:
    print(f"{col}: {df[col].mode().tolist()}")
```
